```python
import requests
from bs4 import BeautifulSoup
ur1 = "https://tsigmjcet.in"
page = requests.get(ur1)
soup = BeautifulSoup(page.text,"html.parser")
print("first heading")
print(soup.find("h2").text)
print("all heading on the page")
for h in soup.find_all("h2"):
    print("-",h.text)
```
